```python
# Import Libraries
import pandas as pd
import glob
import os

# Function to Load Dataset
def load_dataset(folder_path):
    """
    Load all CSV files from a folder into a single DataFrame.
    Adds a 'subject' column based on file name.
    """
    all_files = glob.glob(os.path.join(folder_path, "*.csv"))
    df_list = []
    
    for file in all_files:
        df = pd.read_csv(file)
        # Add a 'subject' column
        df['subject'] = os.path.basename(file).split('.')[0]
        df_list.append(df)
    
    return pd.concat(df_list, ignore_index=True)


# Label Dictionaries
harth_labels = {
    1: "walking",
    2: "running",
    3: "shuffling",
    4: "stairs (ascending)",
    5: "stairs (descending)",
    6: "standing",
    7: "sitting",
    8: "lying",
    13: "cycling (sit)",
    14: "cycling (stand)",
    130: "cycling (sit, inactive)",
    140: "cycling (stand, inactive)"
}


# Set Dataset Paths
harth_path = './harth/'                 


# Load Each Dataset Separately
harth_df = load_dataset(harth_path)


# Add Dataset + Activity Columns
harth_df['dataset'] = 'HARTH'

harth_df['activity'] = harth_df['label'].map(harth_labels)


# Inspect DataFrames
print("HARTH Dataset:")
display(harth_df.head())



# Subject Counts
print("HARTH subjects:", harth_df['subject'].nunique())


# Activity Counts
print("\nHARTH activity counts:")
print(harth_df['activity'].value_counts())

```

# Windowing and feature extraction
This process takes the continuous IMU time-series data and breaks it into fixed-length windows (every 2 seconds). Within each window, it summarizes the raw signals by calculating simple statistics such as the mean and standard deviation for each sensor axis. This reduces millions of raw samples into compact feature vectors that still capture the key motion characteristics.

```python
# Convert the 'timestamp' column to datetime objects if it's not already
harth_df['timestamp'] = pd.to_datetime(harth_df['timestamp'])

# in order to calculate the difference between consecutive timestamps,
# diff() calculates the difference between the current row and the previous row,
# and returns a timedelta object which represents that duration.

time_diffs = harth_df.groupby('subject')['timestamp'].diff().dt.total_seconds()

# Drop the first row for each subject as the diff will be NaN
time_diffs = time_diffs.dropna()

# Calculate the average time difference
average_time_diff = time_diffs.mean()

# The sampling rate is 1 divided by the average time difference
calculated_sampling_rate = 1 / average_time_diff

print(f"Average time difference between samples: {average_time_diff:.4f} seconds")
print(f"Calculated sampling rate: {calculated_sampling_rate:.2f} Hz")
```

```python
# ~2-second windows if sampling ~50Hz => 100 samples per window
window_size = 100
harth_filtered = harth_filtered.sort_values(['subject','timestamp']).reset_index(drop=True)
harth_filtered['window_id'] = harth_filtered.groupby('subject').cumcount() // window_size

windowed = (
    harth_filtered
      .groupby(['subject','activity','window_id'])[sensor_cols]
      .agg(['mean','std'])
      .reset_index()
)

print("\nWindowed dataset preview")
print(windowed.head(8))
```

```python
# Shape of the windowed dataset
print("Windowed dataset shape (HARTH):", windowed.shape)

# Number of rows (windows)
print("Number of windows (HARTH):", len(windowed))

# Number of unique subjects and activities
print("Unique subjects (HARTH):", windowed['subject'].nunique())
print("Unique activities (HARTH):", windowed['activity'].nunique())

# Rows per activity (helps see balance)
print("\nWindow counts per activity (HARTH):")
print(windowed['activity'].value_counts().to_string())
```

# Computing RMS and Harmonic Ration (HR)

RMS measures overall movement intensity across the back and thigh sensors. A higher RMS means stronger, more vigorous steps and a lower RMS suggests lighter or shuffling movement. HR is a smoothness score computed from the trunk (back) signal in the frequency domain. A higher HR means gait is more regular and symmetric, while a lower HR indicates choppier, less stable walking. The code prints a few window-level examples to sanity-check the numbers and then shows per-subject averages, giving a quick profile of each person’s typical step intensity (RMS) and smoothness (HR).

```python
import numpy as np
import pandas as pd

FS = 50  # Hz
EPS = 1e-8

# helpers
def window_rms(df):
    """Compute overall movement intensity across sensors in one window."""
    cols = ['back_x','back_y','back_z','thigh_x','thigh_y','thigh_z']
    x = df[cols].to_numpy(float)
    return float(np.sqrt((x**2).mean()))

def harmonic_ratio(z, fs=FS, fmin=0.5, fmax=3.0):
    """Compute gait smoothness (HR) from trunk vertical acceleration."""
    z = np.asarray(z, float)
    if len(z) < 30:
        return np.nan
    z -= z.mean()
    Y = np.fft.rfft(z)
    f = np.fft.rfftfreq(len(z), 1/fs)
    band = (f >= fmin) & (f <= fmax)
    if not band.any():
        return np.nan
    mag2 = (np.abs(Y)**2)
    pk_idx = np.where(band)[0][np.argmax(mag2[band])]
    f0 = f[pk_idx]
    if f0 <= 0:
        return np.nan
    max_h = int(f[-1] // f0)
    if max_h < 2:
        return np.nan
    df = fs/len(z)
    win = 1.5*df
    evenE = oddE = 0.0
    for h in range(1, max_h+1):
        mask = (f >= h*f0 - win) & (f <= h*f0 + win)
        e = mag2[mask].sum()
        if h % 2 == 0:
            evenE += e
        else:
            oddE += e
    return np.nan if oddE <= 0 else float(evenE/(oddE + EPS))

# compute (walking only)
walk = harth_filtered[harth_filtered['activity'] == 'walking'].copy()

rows = []
for (subj, win), g in walk.groupby(['subject','window_id']):
    rms = window_rms(g)
    hr  = harmonic_ratio(g['back_z'].values)
    rows.append({'subject': subj, 'window_id': win, 'rms': rms, 'hr': hr})

metrics = pd.DataFrame(rows).sort_values(['subject','window_id'])

# print samples
print("\nPer-window (first 10):")
print(metrics.head(10).to_string(index=False))

# per-subject means (simple summary)
summary = (metrics.groupby('subject')[['rms','hr']].mean()
                  .rename(columns={'rms':'rms_mean','hr':'hr_mean'})
                  .reset_index())

print("\nPer-subject means:")
print(summary.to_string(index=False))
```

# Build window features & labels for model training (RF)
This block converts raw IMU samples into one feature row per 2-second window (per subject) and a single label for that window. First, it assigns each window a majority-vote activity (so brief transitions don’t flip the label). Then it summarizes signal shape by computing mean and standard deviation for each sensor axis (back_x/y/z, thigh_x/y/z). To capture overall movement, it also forms vector magnitudes for trunk (back_mag) and thigh (thigh_mag) and takes their mean/std per window. These summaries are merged into Xw, which still includes subject and window_id for proper grouped splits later; the aligned labels live in yw. Finally, Xw_model drops the identifier columns so the Random Forest trains only on features (not subject IDs), and the prints confirm feature matrix size and class distribution.

```python
import numpy as np

def build_window_features(df, axes):
    """
    df must contain: subject, window_id, activity, and sensor axes in `axes`.
    Returns:
      Xw  : DataFrame of features with subject/window_id columns included
      yw  : Series of majority-vote labels per (subject, window_id)
    """
    # Majority-vote label per window (handles transitions when they occur)
    y_win = (df.groupby(['subject','window_id'])['activity']
               .agg(lambda s: s.mode().iat[0]))

    # Mean/std per axis
    feats = (df.groupby(['subject','window_id'])[axes].agg(['mean','std']))
    feats.columns = [f'{a}_{b}' for a,b in feats.columns]
    feats = feats.reset_index()

    # Vector magnitudes (back & thigh) mean/std
    back_mag  = np.sqrt(df['back_x']**2 + df['back_y']**2 + df['back_z']**2)
    thigh_mag = np.sqrt(df['thigh_x']**2 + df['thigh_y']**2 + df['thigh_z']**2)
    df2 = df.assign(back_mag=back_mag, thigh_mag=thigh_mag)
    mag = (df2.groupby(['subject','window_id'])[['back_mag','thigh_mag']].agg(['mean','std']))
    mag.columns = [f'{a}_{b}' for a,b in mag.columns]
    mag = mag.reset_index()

    X = feats.merge(mag, on=['subject','window_id'], how='left')
    y = y_win.loc[pd.MultiIndex.from_frame(X[['subject','window_id']])].reset_index(drop=True)
    return X, y

# Build from your harth_filtered (already windowed via window_id)
Xw, yw = build_window_features(harth_filtered, sensor_cols)

# Keep subject IDs for grouped splits, and strip index cols from the feature matrix
groups = Xw['subject'].values
Xw_model = Xw.drop(columns=['subject','window_id'])

print("\n[Features] X shape:", Xw_model.shape, "| y length:", len(yw))
print("[Features] Classes:", sorted(yw.unique().tolist()))
print("[Features] Sample rows:\n", Xw_model.head(5).to_string(index=False))
```

# Using Thigh Features Only
Yields 0.854 accuracy.

```python
# ===============================
# === THIGH-only Random Forest (subject-wise split)
# ===============================

import time
import matplotlib.pyplot as plt
from sklearn.model_selection import GroupShuffleSplit
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, ConfusionMatrixDisplay, accuracy_score
from sklearn.ensemble import RandomForestClassifier
import joblib
import pandas as pd  # make sure this is imported somewhere above

# Only use thigh features
thigh_cols = [
    'thigh_x_mean', 'thigh_x_std',
    'thigh_y_mean', 'thigh_y_std',
    'thigh_z_mean', 'thigh_z_std',
    'thigh_mag_mean', 'thigh_mag_std'
]

print(f"[Cols] Using {len(thigh_cols)} thigh-only features.")

# Subject-wise split (70/30-ish since we're only doing train/test here)
gss_th = GroupShuffleSplit(n_splits=1, test_size=0.15, random_state=42)
trainval_idx, test_idx = next(gss_th.split(Xw_model[thigh_cols], yw, groups=groups))

X_trainval_th = Xw_model.iloc[trainval_idx][thigh_cols].copy()
X_test_th     = Xw_model.iloc[test_idx][thigh_cols].copy()
y_trainval    = yw.iloc[trainval_idx]
y_test        = yw.iloc[test_idx]

print(f"\n[THIGH Split] Train/Val: {X_trainval_th.shape}, Test: {X_test_th.shape}")
print("[THIGH Split] Train/Val class distribution:\n",
      y_trainval.value_counts().to_string())
print("[THIGH Split] Test class distribution:\n",
      y_test.value_counts().to_string())

# Thigh-only RF model (hyperparams fixed here; no GridSearch dependency)
rf_thigh = Pipeline([
    ('clf', RandomForestClassifier(
        n_estimators=400,
        max_depth=None,
        min_samples_leaf=1,
        class_weight='balanced',
        random_state=42,
        n_jobs=-1
    ))
])

print("[RF-THIGH] fitting...")
t0 = time.perf_counter()
rf_thigh.fit(X_trainval_th, y_trainval)
fit_time = time.perf_counter() - t0

y_pred_th = rf_thigh.predict(X_test_th)
acc_th = accuracy_score(y_test, y_pred_th)

print(f"[RF-THIGH] fit time: {fit_time:.2f}s")
print(f"[RF-THIGH] Test Accuracy: {acc_th:.3f}")
print("\n[RF-THIGH] Classification report (held-out subjects):")
print(classification_report(y_test, y_pred_th, digits=3))

fig, ax = plt.subplots(figsize=(7,6))
ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_th, ax=ax, xticks_rotation=45, colorbar=False
)
ax.set_title("HARTH — Random Forest (THIGH-only features, held-out subjects)")
plt.tight_layout()
plt.show()

# ===============================
# Save THIGH-only estimator + feature schema
# ===============================

joblib.dump(rf_thigh, 'harth_rf_thigh.pkl')
print("Saved THIGH-only model -> harth_rf_thigh.pkl")

thigh_feature_names = X_trainval_th.columns
pd.Series(thigh_feature_names).to_csv('thigh_feature_names.csv',
                                      index=False, header=False)
print("Saved THIGH feature schema -> thigh_feature_names.csv")
```

# Top configurations per tree count and heatmaps
Quick “leaderboard” of the best depth/leaf combos within each n_estimators, so we can see which structure consistently performs well. Visualize how max_depth and min_samples_leaf interact at each n_estimators. This highlights stability of good regions (e.g., shallow vs. deep trees) across tree counts.

```python
# Top configs per tree count
import pandas as pd

K = 5  # top K per n
cvres = pd.DataFrame(rf_gs.cv_results_)

tops = []
for n in sorted(cvres['param_clf__n_estimators'].unique()):
    sub = cvres[cvres['param_clf__n_estimators'] == n]
    bestk = sub.nlargest(K, 'mean_test_score')[['param_clf__max_depth',
                                                'param_clf__min_samples_leaf',
                                                'mean_test_score','mean_fit_time']]
    bestk = bestk.assign(n_estimators=n)
    tops.append(bestk)

tops = pd.concat(tops, ignore_index=True)
display(tops.style.format({'mean_test_score':'{:.3f}','mean_fit_time':'{:.2f}'}))
```

```python
# Heatmaps per n_estimators: depth x leaf performance
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

cvres = pd.DataFrame(rf_gs.cv_results_)
ns = sorted(cvres['param_clf__n_estimators'].unique())
cols = len(ns)
fig, axes = plt.subplots(1, cols, figsize=(5*cols, 4), constrained_layout=True, sharey=True)

for i, n in enumerate(ns):
    sub = cvres[cvres['param_clf__n_estimators'] == n]
    pivot = sub.pivot_table(index='param_clf__max_depth',
                            columns='param_clf__min_samples_leaf',
                            values='mean_test_score',
                            aggfunc='mean')
    ax = axes[i] if cols > 1 else axes
    im = ax.imshow(pivot.values, aspect='auto', origin='lower', cmap='viridis')
    ax.set_title(f'n_estimators={n}')
    ax.set_xlabel('min_samples_leaf')
    if i == 0: ax.set_ylabel('max_depth')
    ax.set_xticks(range(pivot.shape[1])); ax.set_xticklabels(pivot.columns.astype(str))
    ax.set_yticks(range(pivot.shape[0])); ax.set_yticklabels(pivot.index.astype(str))
    fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04, label='CV macro-F1')

plt.show()
```

Across our tuning runs, macro-F1 is essentially flat (~0.881) as we increase the number of trees from 200 to 600, while training time grows nearly linearly (≈37s → 75s → 110s), so adding trees yields diminishing returns. The grid-search table shows the best scores (≈0.889–0.891 macro-F1) occur with deeper trees (around max_depth=24) and small leaves (min_samples_leaf=1–2); unlimited depth performs similarly but is slower and risks overfitting, and larger leaves (≥4) consistently hurt performance. The heatmaps reinforce this: brighter regions cluster at deeper depths with smaller leaves, and the patterns are almost identical across 200/400/600 trees, confirming that more trees don’t materially change accuracy. Overall, a fast, strong configuration for this dataset is n_estimators=200, max_depth=24, min_samples_leaf=1–2, class_weight='balanced', which captures the best accuracy region while minimizing training time.
